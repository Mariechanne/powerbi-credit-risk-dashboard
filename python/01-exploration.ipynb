## 1. Premier aperçu de la structure
---

Le fichier complet contient plus de deux millions de prêts. Pour découvrir sa structure sans saturer la mémoire, on lit d'abord un échantillon des 1 000 premières lignes : il suffit pour connaître le nombre de colonnes et leurs noms.

```python
import pandas as pd

chemin = "../data/raw/accepted_2007_to_2018Q4.csv.gz"
apercu = pd.read_csv(chemin, nrows=1000)

print(f"Echantillon : {apercu.shape[0]} lignes et {apercu.shape[1]} colonnes")

# Un tableau qui décrit chaque colonne
description = pd.DataFrame({
    "type": apercu.dtypes.astype(str),
    "manquants (%)": (apercu.isna().mean() * 100).round(1),
    "exemple": apercu.apply(lambda col: col.dropna().iloc[0] if col.notna().any() else None),
})

pd.set_option("display.max_rows", 200)
description
```
